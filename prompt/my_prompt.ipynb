```python
import os
# ધારી લો કે તમે યોગ્ય AI લાઇબ્રેરી ઇન્સ્ટોલ અને સેટઅપ કરી છે
# e.g., from google import genai

# ૧. ટેસ્ટ માટેનું ફિક્સ ઇનપુટ (પ્રોડક્ટની માહિતી)
product_input = "Eco-Friendly Bamboo Water Bottle, 750ml, keeps water cold for 12 hours, stainless steel interior."

# ૨. ત્રણેય અલગ-અલગ પ્રોમ્પટ્સ
prompt_1_vague = f"Write a product description for this: {product_input}"

prompt_2_clear = f"Write a short and catchy product description (2-3 sentences) for an e-commerce website based on these details: {product_input}. Focus on health and environment benefits."

prompt_3_strong = f"""
System: You are an expert e-commerce copywriter who writes high-converting, punchy product descriptions.

Task: Write a product description based on the provided details. Follow the style of the example below.

Example Input: Smart LED Bulb, Wi-Fi enabled, 16 million colors, works with Alexa.
Example Output: 
- **Headline**: Light Up Your Life, Your Way!
- **Description**: Experience ultimate convenience with our Smart LED Bulb. Choose from 16 million colors and control your ambiance with simple voice commands via Alexa.
- **Key Feature**: Smart Wi-Fi Integration.

Current Input: {product_input}
Output Format:
- **Headline**: [Catchy Headline]
- **Description**: [2 sentences description]
- **Key Feature**: [Main Benefit]
"""

# ૩. પ્રોમ્પટ્સને લિસ્ટમાં સેવ કરવા
prompts_to_test = {
    "1. Vague Prompt": prompt_1_vague,
    "2. Clear Prompt": prompt_2_clear,
    "3. Strong Prompt": prompt_3_strong
}

# ૪. લૂપ ચલાવીને ટેસ્ટ કરવું અને આઉટપુટ સેવ કરવું
results = {}

print("--- પ્રોમ્પટ ટેસ્ટિંગ શરૂ થઈ રહ્યું છે ---\n")
for name, prompt_text in prompts_to_test.items():
    print(f"Running {name}...")
    
    # અહીં AI મોડલને કોલ કરવો (કાલ્પનિક ફંક્શન call_ai_model)
    # response = call_ai_model(prompt_text)
    # response_text = response.text
    
    # સમજવા માટે ટેમ્પરરી આઉટપુટ પ્લેસહોલ્ડર
    response_text = f"[AI Response for {name}]" 
    
    results[name] = {
        "prompt": prompt_text,
        "output": response_text
    }

# ૫. પરિણામોને ફાઈલમાં સેવ કરવા (JSON અથવા Text)
import json
with open("prompt_test_results.json", "w", encoding="utf-8") as f:
    json.dump(results, f, indent=4, ensure_ascii=False)

print("\nબધા પ્રોમ્પટ્સ અને તેમના આઉટપુટ 'prompt_test_results.json' ફાઈલમાં સેવ થઈ ગયા છે!")
```
