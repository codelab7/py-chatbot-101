```python
import pandas as pd
data = {
    "name" : ["A","B","C"],
    "Marks" : [20,15,30]
}


df=pd.DataFrame(data)
print(df)

#   name  Marks
# 0    A     20
# 1    B     15
# 2    C     30

A=df.head()
print("Student data:",A)

# Student data:   name  Marks
# 0    A     20
# 1    B     15
# 2    C     30


#particular colums Select
df["Marks"]
# 0    20
# 1    15
# 2    30
# Name: Marks, dtype: int64




#Filter 
b = df[df["Marks"] >15]
print(b)
#   name  Marks
# 0    A     20
# 2    C     30
```
