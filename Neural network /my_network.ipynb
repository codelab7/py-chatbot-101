**What is a neuron in a neural network** 

neuron e network no nano part che. je ak nanu decision maker che.

tenu work:  
Input levu
Important information ne vadhare importance apvi 
Decision levu
Output apvu

**Example** :

Maan lo phone face unlock kari rahyu che.
Neuron check kari shake:
Eye shape
Nose position
Mouth shape
Face distance

Each neuron ek nani detail joi ne decision aape che.

**Neuron kevi rite work kare**

Neuron ma mainly 3 part hoy che. 

1. input
Aa information hoy che.

Example:
Height
Weight
Pixel values
Face features

2. Weights
Weights decide kare che kai input ketlu important che.

Example:
Face unlock ma:
Eyes → more important
Hair style → less important

So eyes nu weight vadhu hase.

3. Output
Aa final decision maker che.

Simple Flow: 
Inputs → Multiply with weights → Add together → Activation → Output

**Layer**

Neural network ma layers atle neuron na group. different layer different work kare che.
simple:
 layer 1 : basic information
 layer 2 : pattern learn kare che 
 layer 3 : final decision le che 

**Example** 

Mano k koi kids car olakhata sikhe che.

layer 1 : basic information le che. jem k car wheel che k nai e jove che.

layer 2 : pachi samje k car kevi che. car no shape jove 

layer 3 : pachi final decision le che k aa car che.

ak layer thi khali simple pattern j sikhi sakhay che. pn jo multiple layer hoy to comple patterns,deep understanding and better prediction kari sakhay che. atle mate multiple layer ni Important che.
 

**During Training** : 

Training atle neural network ne example apje ne shikhavani process che. Network start ma correct ans nathi apto. te guess kare che, mistake kare che and pachi e mistake thi shikhe che. te prediction kare che , loss calculate kare che and backpropagation thi weights update kare che and pachi better predicition karta shikhe che.

**Loss Function**

Loss function e neural network ni mistake mapvanu kam kare che.

te check kare che k model na ans actual ans ma ketlo difference che.
jo loss vadhu to prediction wrong hoy and jo loss ocho to prediction vadhare hoy che.

Training:
Network prediction kare che
Actual answer thay compare thay che
Loss function mistake calculate kare che
Network weights update kare che

Loss function e model ni prediction ketli worn che te measure karvani function che.
