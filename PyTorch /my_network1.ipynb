```python
import torch
import torch.nn as nn

# features input data 
house_size = 3000.0
bedrooms = 4.0
house_age = 5.0

# input data convert tensor
house = torch.tensor([[house_size, bedrooms, house_age]])

# model
model = nn.Linear(in_features=3, out_features=1)

# ૩. forward pass
with torch.no_grad(): 
    predicted_price = model(house)

print("Predict Houseprice:", predicted_price.item())
```
