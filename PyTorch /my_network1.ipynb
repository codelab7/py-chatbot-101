```python
import torch
import torch.nn as nn


# ૧. ઇનપુટ ડેટા વ્યાખ્યાયિત કરો (Tensor)
x = torch.tensor([[1.0, 2.0]])

# ૨. એક નાનું ન્યુરલ નેટવર્ક મોડલ બનાવો
model = nn.Sequential(
    nn.Linear(2,3),   # લીનિયર લેયર (2 ઇનપુટ ફીચર્સ -> 3 આઉટપુટ ફીચર્સ)
    nn.ReLU()          # એક્ટિવેશન ફંક્શન
)

# ૩. ફોરવર્ડ પાસ કરો (Forward Pass)
output = model(x)

# ૪. આઉટપુટ પ્રિન્ટ કરો
print(output)
```

```python
import torch
import torch.nn as nn


# ૧. ઇનપુટ ડેટા વ્યાખ્યાયિત કરો (Tensor)
x = torch.tensor([1.0],requires_grad=True)

# ૨. એક નાનું ન્યુરલ નેટવર્ક મોડલ બનાવો
model = nn.Sequential(
    nn.Linear(1,1),   # લીનિયર લેયર (2 ઇનપુટ ફીચર્સ -> 3 આઉટપુટ ફીચર્સ)
          
)

# ૩. ફોરવર્ડ પાસ કરો (Forward Pass)
output = model(x)

# ૪. આઉટપુટ પ્રિન્ટ કરો
print(output.item())
```
