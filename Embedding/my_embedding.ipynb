```python
from sentence_transformers import SentenceTransformer, util

# 1. મોડલ લોડ કરો
model = SentenceTransformer('paraphrase-multilingual-MiniLM-L12-v2')

# 2. વાક્યો વ્યાખ્યાયિત કરો
sentences = [
    "I like programming.",
    "I like coding.",
    "The weather is lovely today."
]

# 3. એમ્બેડિંગ્સ મેળવો
embeddings = model.encode(sentences)

for i, sentence in enumerate(sentences):
    # દરેક વાક્યના એમ્બેડિંગ વેક્ટરના શરૂઆતના ૫ નંબર્સ સેમ્પલ તરીકે જોવા માટે
    print(f"sentences: '{sentence}'")
    print(f"Shape: {embeddings[i].shape} ")
    print(f"Sample Vector: {embeddings[i][:5]}")


sim_1_2 = util.cos_sim(embeddings[0], embeddings[1])
sim_1_3 = util.cos_sim(embeddings[0], embeddings[2])


print("Cosine similarity")
print(f"Sentences 1 and 2: {sim_1_2.item():.4f}")
print(f"Sentences 1 and 3 {sim_1_3.item():.4f}")
```
