```python
# Import libraries
import pandas as pd
from sklearn.model_selection import train_test_split


# Load dataset
df = pd.read_csv("train.csv")

# Features (input data)
X = df[['Pclass', 'Age', 'Fare']]

# Label (target)
y = df['Survived']

X = X.dropna()
y = y[X.index]

# Split data into training set (80%) and test set (20%)
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,   # 20% for testing
    random_state=10   # same split every time
)

# Print sizes
print("Training data size:", X_train.shape)
print("Testing data size:", X_test.shape)

print("Trainning data : ")
print(X_train)
print(y_train)

print("Testing Data : ")
print(X_test)
print(y_test)

```

**Train data and Test data on same data** : If you train and test on exactly the same data, the model will usually give very high accuracy — but this can be                                             misleading.Because model already saw those answers during training. It becomes like memorizing, not real                                                     learning.
