```python
#Sentiment analysis

from transformers import pipeline, logging
import warnings

# ૧. પાયથોનની સ્ટાન્ડર્ડ વોર્નિંગ્સ બંધ કરવા માટે
warnings.filterwarnings("ignore")

# ૨. હગિંગ ફેસ (Transformers) ના ઇન્ટરનલ મેસેજ અને પ્રોગ્રેસ બાર છુપાવવા માટે
logging.set_verbosity_error()

# ૩. અહીં મોડલનું નામ સ્પષ્ટ લખવાથી "No model was supplied" વાળી વોર્નિંગ દૂર થશે
classifier = pipeline("sentiment-analysis")


# ૧. પોઝિટિવ વાક્યનું ઉદાહરણ
text1 = "I absolutely love coding in Python! It is amazing."
result1 = classifier(text1)

print("=== Example 1===")
print(f"Sentences: '{text1}'")
print(f"Result : {result1[0]['label']}")
print(f"Score: {result1[0]['score']:.4f}\n")

# ૨. નેગેટિવ વાક્યનું ઉદાહરણ
text2 = "I am not coding person"
result2 = classifier(text2)

print("=== Example 2 ===")
print(f"Sentences: '{text2}'")
print(f"Result: {result2[0]['label']}")
print(f"Score: {result2[0]['score']:.4f}\n")
```

```python
import os
# Hide Hugging Face token and symlink related warnings
os.environ["HF_HUB_DISABLE_SYMLINKS_WARNING"] = "1"
os.environ["TRANSFORMERS_NO_ADVISORY_WARNINGS"] = "true"

from transformers import pipeline, logging
import warnings

# 1. Disable standard Python warnings
warnings.filterwarnings("ignore")

# 2. Hide internal Hugging Face messages and progress bars
logging.set_verbosity_error()
# 3. Text Generation પાઇપલાઇન તૈયાર કરવી (આમાં ડિફોલ્ટ gpt2 મોડલ લોડ થશે)
generator = pipeline("text-generation")



# 1. Example 1: Starting a technology topic
text1 = "In the future, Artificial Intelligence will"
# max_length: આઉટપુટમાં ટોટલ કેટલા શબ્દો રાખવા છે તે સંખ્યા
result1 = generator(text1, max_length=40, num_return_sequences=1)

print("=== Example 1 ===")
print(f"Sentences: '{text1}'")
print(f"Result : {result1[0]['generated_text']}\n")

# # 2. Example 2: Starting a programming topic
# text2 = "Python is the best programming language because"
# result2 = generator(text2, max_length=40, num_return_sequences=1)

# print("=== Example 2 ===")
# print(f"Sentences: '{text2}'")
# print(f"Result: {result2[0]['generated_text']}\n")
```
