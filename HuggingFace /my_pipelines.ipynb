```python
#Sentiment analysis

from transformers import pipeline
import warnings

# ૧. પાયથોનની સ્ટાન્ડર્ડ વોર્નિંગ્સ બંધ કરવા માટે
warnings.filterwarnings("ignore")


# ૩. અહીં મોડલનું નામ સ્પષ્ટ લખવાથી "No model was supplied" વાળી વોર્નિંગ દૂર થશે
classifier = pipeline("sentiment-analysis")


# # ૧. પોઝિટિવ વાક્યનું ઉદાહરણ
text1 = "I absolutely love coding in Python! It is amazing."
result1 = classifier(text1)

print("=== Example 1===")
print(f"Sentences: '{text1}'")
print(f"Result : {result1[0]['label']}")
print(f"Score: {result1[0]['score']:.4f}\n")

# ૨. નેગેટિવ વાક્યનું ઉદાહરણ
text2 = "I am not happy person."
result2 = classifier(text2)

print("=== Example 2 ===")
print(f"Sentences: '{text2}'")
print(f"Result: {result2[0]['label']}")
print(f"Score: {result2[0]['score']:.4f}\n")
```

```python
from transformers import pipeline

# આપણે Google નું t5-base મોડલ વાપરીશું, જે ટેક્સ્ટ સમજીને પોતાની રીતે લખે છે
summarizer = pipeline("summarization", model="t5-base")

article = """
Chocolate is loved by people all over the world. It is made from cacao seeds, which grow on trees in warm places. 
Milk chocolate is sweet and smooth because it has lots of sugar and milk. 
On the other hand, dark chocolate is a bit better but it is very healthy for our heart. 
Doctors say eating a little bit of dark chocolate every day can make us happy and energetic.
"""

# મોડલને પોતાની રીતે વિચારવા દેવા માટે 'do_sample=True' અને 'temperature' સેટ કર્યું
summary = summarizer(article, max_length=60, min_length=30, do_sample=True, temperature=0.7)

print("--summary text--")
print(summary[0]['summary_text'])
```

```python

from transformers import pipeline

# ૧. ટ્રાન્સલેશન માટે પાઈપલાઈન બનાવો
translator = pipeline("translation", model="facebook/nllb-200-distilled-600M")

# ૨. તમારો ઇંગ્લિશ પેરેગ્રાફ (ઇનપુટ)
english_text = "Chocolate is loved by people all over the world. It is very healthy for our heart."

# ૩. પાઈપલાઈન રન કરો 
result = translator(english_text, src_lang="eng_Latn", tgt_lang="guj_Gujr")

# ૪. ફાઇનલ ગુજરાતી આઉટપુટ પ્રિન્ટ કરો
print(result[0]['translation_text'])
```

```python
from transformers import pipeline

# ૧. ક્વેશ્ચન-આન્સરિંગ માટે પાઈપલાઈન બનાવો
qa_predictor = pipeline("question-answering", model="deepset/roberta-base-squad2")

# ૨. આપણો પેરેગ્રાફ (Context)
paragraph = """
Chocolate is loved by people all over the world. It is made from cacao seeds, which grow on trees in warm places. 
Milk chocolate is sweet and smooth because it has lots of sugar and milk. 
On the other hand, dark chocolate is a bit bitter but it is very healthy for our heart.
"""

# ૩. આપણો સવાલ (Question)
my_question = "What is Milk chocolate?why?"

# ૪. પાઈપલાઈન રન કરો (અહીં સવાલ અને પેરેગ્રાફ બંને એકસાથે ડિક્શનરીમાં પાસ કરવા પડે)
result = qa_predictor(question=my_question, context=paragraph)

# ૫. જવાબ પ્રિન્ટ કરો
print("Answer : ",result['answer'])
```
