```python
#Sentiment analysis

from transformers import pipeline, logging
import warnings

# ૧. પાયથોનની સ્ટાન્ડર્ડ વોર્નિંગ્સ બંધ કરવા માટે
warnings.filterwarnings("ignore")

# ૨. હગિંગ ફેસ (Transformers) ના ઇન્ટરનલ મેસેજ અને પ્રોગ્રેસ બાર છુપાવવા માટે
logging.set_verbosity_error()

# ૩. અહીં મોડલનું નામ સ્પષ્ટ લખવાથી "No model was supplied" વાળી વોર્નિંગ દૂર થશે
classifier = pipeline("sentiment-analysis")


# ૧. પોઝિટિવ વાક્યનું ઉદાહરણ
text1 = "I absolutely love coding in Python! It is amazing."
result1 = classifier(text1)

print("=== Example 1===")
print(f"Sentences: '{text1}'")
print(f"Result : {result1[0]['label']}")
print(f"Score: {result1[0]['score']:.4f}\n")

# ૨. નેગેટિવ વાક્યનું ઉદાહરણ
text2 = "I am not coding person"
result2 = classifier(text2)

print("=== Example 2 ===")
print(f"Sentences: '{text2}'")
print(f"Result: {result2[0]['label']}")
print(f"Score: {result2[0]['score']:.4f}\n")
```

```python
from transformers import pipeline

# ૧. શુદ્ધ 'translation' ટાસ્ક અને ગુગલનું ઓફિશિયલી સપોર્ટેડ મોડલ વાપરો
# અહીં આપણે English થી French (અથવા German, Romanian) માં ઓફિશિયલ ટ્રાન્સલેશન કરી શકીશું
translator = pipeline("translation_en_to_fr", model="t5-small")

# ૨. તમારો અંગ્રેજી ટેક્સ્ટ
english_text = "I love coding and building intelligent software."

# ૩. માત્ર પ્યોર પાઇપલાઇન રન કરો (કોઈ બીજા લોચા વગર)
output = translator(english_text)

# ૪. આઉટપુટ પ્રિન્ટ કરો
print("Original:", english_text)
print("Translation:", output[0]['translation_text'])
```

```python
from transformers import pipeline

# ૧. શુદ્ધ 'text2text-generation' ટાસ્ક અને ગુગલનું ગુજરાતી સપોર્ટેડ મોડલ લોડ કરો
translator = pipeline("text2text-generation", model="google/mt5-small")

# ૨. તમારો અંગ્રેજી ટેક્સ્ટ (આમાં મોડલને ચોખ્ખું કહેવું પડે કે ગુજરાતીમાં અનુવાદ કર)
english_text = "translate English to Gujarati: I love Coding."

# ૩. પ્યોર પાઇપલાઇન રન કરો
output = translator(english_text, max_new_tokens=30)

# ૪. આઉટપુટ પ્રિન્ટ કરો
print(output[0]['generated_text']) 
```
